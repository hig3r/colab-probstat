# 準備



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
%matplotlib inline
```

```python
np.random.seed(1637)
```

# カイ二乗分布


$Z_1,\dots,Z_k, \mathrm{i.i.d.}\sim\mathrm{N}(0,1^2)$ とする．

$W_k=Z_1^2+\cdots+Z_k^2$ は， 自由度$k$ のカイ二乗分布にしたがうといい， $W_k\sim\chi_k^2$ とかく．

```python
rvw1=stats.chi2(df=1) # 自由度1
rvw3=stats.chi2(df=3) # 自由度3
```

## 母平均値・母分散・母標準偏差の計算

```python
rvw3.mean() # 母平均値 mean
```

```python
rvw3.var()
```

## 確率密度関数 $f_W(w)$ の計算

```python
rvw3.pdf(1) # f(1) pdf=probability density function
```

## 累積分布関数 $F_W(d)=P(W\leq d)$ の計算

```python
rvw3.cdf(3) # P(X<=3)  cdf=cumulative distribution function
```

## 分位数 $F_W^{-1}(1-\alpha/2)$ の計算
分位数 $F^{-1}(1-\alpha/2)=\chi^2_{k,\alpha/2}$.
同じことだけど $P(W\geq \chi^2_{k,\alpha/2})=\alpha/2$.

$\chi^2_{k,0.01/2}$

```python
rvw3.ppf(1-0.005) # P(X<=?)=0.9.  ppf=percent point function
```

## 確率密度関数, 累積分布関数のグラフ

```python
x=np.linspace(-1,6,71) # 点を打つ x の選択
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvw3.pdf(x))
ax.plot(x,rvw3.cdf(x))
ax.plot(x,rvw1.pdf(x))
ax.plot(x,rvw1.cdf(x))
ax.set_xlabel("x")
ax.set_ylabel("f(x), F(x)")
ax.legend(["f_W(x), k=3", "F_W(x) k=3", "f_W(x) k=1", "F_W(x) k=1"])
ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.axhline(1, color='black', linewidth=0.8, linestyle='--')
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')
fig.show()
```
