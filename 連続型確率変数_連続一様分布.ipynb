```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
%matplotlib inline
```

```python
np.random.seed(1637)
```

# 連続型確率変数

scipy.stats で確率密度関数を自由に与えて, 連続型確率変数を扱うのは少したいへん(自分でクラスを継承して定義する必要がある).

代わりに, 多くの有名な連続型確率変数があらかじめ定義されているので, その固有名詞を探し, 母数(パラメタ)を与えて利用する.

例:指数分布 expon, 一様分布 uniform , 正規分布 norm , カイ二乗分布 chi2 , t分布 t, …



## 連続一様分布

連続一様分布のU($c,d$)の確率密度関数

$$ f(x)=
\begin{cases}
\frac{1}{d-c} & (c\leq x \leq d)\\
0 & (\text{他})\\
\end{cases}
$$

```python
#  U(1,3)
c=1
d=3
loc=c
scale=d-c
rvx = stats.uniform(loc=loc,scale=scale)
```

## 母平均値・母分散・母標準偏差の計算

```python
rvx.mean() # 母平均値 mean
```

```python
rvx.var() # 母分散 variance
```

```python
rvx.std() # 母標準偏差 standard deviation
```

## 母期待値・モーメント

```python
rvx.moment(4) # 4次のモーメント
```

```python
rvx.expect(lambda x: x**2+5*x+3) # 母期待値 E[X^2+5X+3]
```

## 確率密度関数 $f(x)$ の計算

```python
rvx.pdf(0.2) # f(0.2) pdf=probability density function
```

## 累積分布関数 $F(d)=P(X\leq d)$ の計算

```python
rvx.cdf(0.2) # P(X<=0.2)  cdf=cumulative distribution function
```

## 区間の確率の計算
$ P(x_1 \lt X \leq x_2)=F(x_2)-F(x_1)$

```python
rvx.cdf(1.5)-rvx.cdf(0.5)  # P(0.5<X<=1.5)
```

## 分位数 $F^{-1}(q)$ の計算

```python
rvx.ppf(0.9) # P(X<=?)=0.9.  ppf=percent point function
```

## 確率密度関数, 累積分布関数のグラフ

```python
x=np.arange(0,4,0.01)
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvx.pdf(x))
ax.plot(x,rvx.cdf(x))
ax.hlines(1,0,4)
ax.set_xlabel("x")
ax.set_ylabel("f(x), F(x)")
ax.legend(["f(x)", "F(x)","1"])
fig.show()
```

## 標本ナントカ

## 標本抽出

```python
n=30
samplex=rvx.rvs(size=n)  # rvs=random variable sample
samplex
```

```python
[samplex.mean(),samplex.var(),samplex.std()]
```

### ヒストグラム

```python
# ヒストグラム
fig,ax = plt.subplots()
ax.hist(samplex,bins=7)
ax.set_ylim(0,n)
ax.set_xlim(0,4)
fig.show()
```

```python
# 累積ヒストグラム
fig,ax = plt.subplots()
ax.hist(samplex,bins=n,cumulative=True,histtype='stepfilled')
ax.set_ylim(0,n)
ax.set_xlim(0,4)
fig.show()
```

# 連続型確率変数の1次変換
$Y=aX+b$

例
$X\sim \mathrm{U}(1,3)$
$Y=2X+1$
ならば
$Y\sim \mathrm{U}(3,7)$


```python
a=2
b=1
```

### 母ナントカ

```python
rvy=stats.uniform(loc=a*c+b,scale=a*d-a*c)
```

```python
x=np.arange(0,8,0.01)
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvx.pdf(x))
ax.plot(x,rvy.pdf(x))
ax.plot(x,rvx.cdf(x))
ax.plot(x,rvy.cdf(x))
ax.set_xlabel("x,y")
ax.set_ylabel("f,F")
ax.legend(["f_X", "f_Y","F_X","F_Y"])
fig.show()
```

### 標本ナントカの変換

```python
sampley=a*samplex+b
sampley
```

```python
# ヒストグラム
fig,ax = plt.subplots()
ax.hist(samplex,bins=5,density=True)
ax.hist(sampley,bins=5,density=True)
ax.set_ylim(0,1)
ax.set_xlim(0,10)
ax.legend(["X","Y"])
fig.show()
```

# 連続型確率変数の標準化
$Z=\frac{X-\mu}{\sigma}$ or $X=\sigma Z+\mu$.

$\mu=b=\mathrm{E}[X], \sigma=a=\sqrt{\mathrm{Var}(X)}$.

例
$X\sim \mathrm{U}(1,3)$
ならば $\mu=2, \sigma=\sqrt{1/3}$.

$Z\sim \mathrm{U}(-\sqrt{3},+\sqrt{3})$

```python
mu=2
sigma=np.sqrt(4/12)
```

### 母ナントカ

```python
rvz=stats.uniform(loc=(c-mu)/sigma,scale=(d-mu)/sigma-(c-mu)/sigma)
```

```python
x=np.arange(-4,4,0.01)
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvx.pdf(x))
ax.plot(x,rvz.pdf(x))
ax.plot(x,rvx.cdf(x))
ax.plot(x,rvz.cdf(x))
ax.set_xlabel("x,z")
ax.set_ylabel("f,F")
ax.legend(["f_X", "f_Z","F_X","F_Z"])
fig.show()
```

### 標本ナントカ

```python
samplez=(samplex-mu)/sigma
samplez
```

```python
# ヒストグラム
fig,ax = plt.subplots()
ax.hist(samplex,bins=5,density=True,alpha=0.5)
ax.hist(samplez,bins=5,density=True,alpha=0.5)
ax.set_ylim(0,1)
ax.set_xlim(-2,4)
ax.legend(["X","Z"])
fig.show()
```
