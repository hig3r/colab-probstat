# 準備



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
%matplotlib inline
```

```python
np.random.seed(1637)
```

# Studentのt分布

$Z\sim\mathrm{N}(0,1^2)$ (標準正規分布)

$W_k\sim\chi_k^2$ (自由度$k$のカイ二乗分布)
とするとき，

$T=\frac{Z}{\sqrt{W_k/k}}\sim \mathrm{t}_k$ (自由度$k$のt分布)


```python
rvt1=stats.t(df=1) # 自由度 degree of freedom=1
rvt5=stats.t(df=5) # 自由度 degree of freedom=5
```

## 母平均値・母分散・母標準偏差の計算

```python
rvt5.mean() # 母平均値 mean
```

```python
rvt5.var()
```

## 確率密度関数 $f(x)$ の計算

```python
rvt5.pdf(1) # f(1) pdf=probability density function
```

## 累積分布関数 $F(d)=P(X\leq d)$ の計算

```python
rvt5.cdf(3) # P(X<=3)  cdf=cumulative distribution function
```

## 分位数 $t_{k,\alpha}$ の計算
$F^{-1}(1-\alpha)=t_{k,\alpha}$
または， $P(T\geq t_{k,\alpha})=\alpha$.

$t_5(0.005)$

```python
rvt5.ppf(1-0.005) # P(T<=?)=0.995.  ppf=percent point function
```

## 確率密度関数, 累積分布関数のグラフ

```python
rvz=stats.norm(loc=0,scale=1) #標準正規分布
```

```python
x=np.linspace(-3,3,61) # 点を打つ x の選択
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvz.pdf(x))
ax.plot(x,rvz.cdf(x))
ax.plot(x,rvt5.pdf(x))
ax.plot(x,rvt5.cdf(x))
ax.plot(x,rvt1.pdf(x))
ax.plot(x,rvt1.cdf(x))
ax.set_xlabel("x")
ax.set_ylabel("f(x), F(x)")
ax.legend(["φ(z)","Φ(z)","f_5(x)", "F_5(x)", "f_1(x)", "F_1(x)"])
ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.axhline(1, color='black', linewidth=0.8, linestyle='--')
ax.axvline(0, color='black', linewidth=0.8, linestyle='--')
fig.show()
```
