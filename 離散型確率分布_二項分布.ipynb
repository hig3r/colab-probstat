```python
import numpy as np
import matplotlib.pyplot as plt
import math as math
from scipy import stats # 確率関係
%matplotlib inline
```

```python
np.random.seed(1637) # 乱数のシードの設定. 乱数が毎回同じになる
```

#二項分布とベルヌイ分布

## ベルヌイ分布 Bin(1,2/5)=Ber(2/5)
$X\sim \mathrm{Bin}(1,\frac25)=\mathrm{Ber}(\frac25)$

### 母期待値

```python
rvx=stats.binom(n=1,p=2.0/5)
```

```python
rvx.mean() # 母平均値
```

```python
rvx.var() # 母分散
```

```python
rvx.std() # 母標準偏差
```

```python
rvx.moment(4) # 4次のモーメント
```

### 確率関数と累積分布関数

```python
x=np.linspace(-1,2,121) # 点を打つ x の選択
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvx.pmf(x))  # 確率関数
ax.plot(x,rvx.cdf(x))   # 累積分布関数
ax.set_xlabel("x")
ax.legend(["f_X(x)", "F_X(x)"])
fig.show()
```

### 標本ナントカ

```python
n=100
samplex=rvx.rvs(size=n) # サンプル
samplex
```

```python
# ヒストグラム
fig,ax = plt.subplots()
ax.hist(samplex,bins=5,alpha=0.5)
ax.set_xlim(-1,2)
ax.legend(["X","Z"])
fig.show()
```

## 二項分布 Bin(10,2/5)
$Y\sim \mathrm{Bin}(10,\frac25)$

```python
rvy=stats.binom(n=10,p=2.0/5)
```

```python
rvy.mean()
```

```python
rvy.var()
```

```python
x=np.linspace(-1,11,121) # 点を打つ x の選択
fig, ax = plt.subplots(1, 1)
ax.plot(x,rvy.pmf(x))  # 確率関数
ax.plot(x,rvy.cdf(x))   # 累積分布関数
ax.set_xlabel("x")
ax.legend(["f_Y(x)", "F_Y(x)"])
fig.show()
```

### $P(Y=5)$

```python
rvy.pmf(5)
```

```python
# math.factorial(5)は階乗 5!
(0.8**5)*(0.2**5)*math.factorial(10)/math.factorial(5)/math.factorial(5)
```

### $P(Y \leq 8)$

```python
rvy.cdf(8)
```

```python
1-(0.8**10)-(0.8**9)*(0.2**1) * math.factorial(10)/math.factorial(9)/math.factorial(1) # 1-p(10)-p(9)
```

### 標本ナントカ

```python
n=100
sampley=rvy.rvs(size=n) # サンプル
sampley
```

```python
# ヒストグラム
fig,ax = plt.subplots()
ax.hist(sampley,bins=13,alpha=0.5)
ax.set_xlim(-1,10+1)
ax.legend(["Y"])
fig.show()
```
